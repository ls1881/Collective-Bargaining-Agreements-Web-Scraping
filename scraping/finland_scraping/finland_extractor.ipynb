```python
import os
import glob
import pytesseract
from pdf2image import convert_from_path

# 1. Define your folders relative to the notebook
pdf_folder = "pdfs"  
txt_folder = "txts"

# Create the txts folder if it doesn't exist
if not os.path.exists(txt_folder):
    os.makedirs(txt_folder)
    print(f"Created folder: '{txt_folder}'")

# NOTE FOR WINDOWS USERS: 
# Uncomment the line below and verify the path if Python cannot find tesseract
# pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

# 2. Find all PDF files inside the 'pdfs' subfolder
search_path = os.path.join(pdf_folder, "*.pdf")
pdf_files = glob.glob(search_path)

if not pdf_files:
    print(f"No PDFs found in '{pdf_folder}'. Please place the Finnish PDFs there.")
else:
    print(f"Found {len(pdf_files)} PDF(s). Starting Finnish OCR conversion...\n")
    
    # 3. Loop through each PDF
    for pdf_file in pdf_files:
        # Extract just the file name (e.g., "tyoehtosopimus")
        base_name = os.path.splitext(os.path.basename(pdf_file))[0]
        
        # Build the exact path for the new text file
        txt_filename = f"{base_name}.txt"
        output_path = os.path.join(txt_folder, txt_filename)
        
        # Skip if we already processed this file (saves huge amounts of time if the script crashes halfway)
        if os.path.exists(output_path):
            print(f"Skipping: {base_name}.txt (Already exists)")
            continue
            
        print(f"OCR Processing: {base_name}.pdf -> {output_path}")
        
        try:
            # Step A: Convert the PDF into a list of high-res images
            # 300 DPI is the standard for reliable OCR
            pages = convert_from_path(pdf_file, dpi=300)
            
            # Step B: Read the text from each image using the Finnish language model
            with open(output_path, "w", encoding="utf-8") as txt_file:
                for i, page_image in enumerate(pages):
                    # lang='fin' is crucial here for Finnish characters and vocabulary
                    text = pytesseract.image_to_string(page_image, lang='fin')
                    
                    if text:
                        txt_file.write(text + "\n")
                        
            print(f"  -> Success: {len(pages)} pages processed.")
                        
        except Exception as e:
            # Catches broken PDFs, missing Poppler, or missing Tesseract
            print(f"  -> [ERROR] Failed to process {base_name}.pdf: {e}")
            
    print("\nAll done! Your Finnish text files are in the 'txts' folder.")
```
